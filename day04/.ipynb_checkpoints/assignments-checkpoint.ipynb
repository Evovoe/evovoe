```python
import numpy as np
```

```python
# Assignment A — Implement ReLU and Sigmoid
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

Z = np.random.randn(3, 3)
print(Z)
print(relu(Z))
print(sigmoid(Z))
```

```python
# Assignment B — Build a 2-Layer Neural Network (Forward Pass Only)
X  = np.random.randn(8, 3)
W1 = np.random.randn(3, 4)
b1 = np.random.randn(4)
W2 = np.random.randn(4, 1)
b2 = np.random.randn(1)

# Layer 1
Z1 = X @ W1 + b1
A1 = relu(Z1)

# Layer 2
Z2 = A1 @ W2 + b2
Y = sigmoid(Z2)

print(Y)
```

```python
# Assignment C — Vector Statistics (AI Version)
def vector_norm(x):
    return np.sqrt(np.sum(x**2))

def normalize_vector(x):
    return x / vector_norm(x)

def cosine_similarity(x, y):
    return np.dot(x, y) / (vector_norm(x) * vector_norm(y))
```
