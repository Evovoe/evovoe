```python
# Assignment A — Implement ReLU and Sigmoid
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

Z = np.random.randn(3, 3)
print(Z)
print(relu(Z))
print(sigmoid(Z))
```

```python
# Assignment B — Build a 2-Layer Neural Network (Forward Pass Only)
X  = np.random.randn(8, 3)
W1 = np.random.randn(3, 4)
b1 = np.random.randn(4)
W2 = np.random.randn(4, 1)
b2 = np.random.randn(1)

# Layer 1
Z1 = X @ W1 + b1
A1 = relu(Z1)

# Layer 2
Z2 = A1 @ W2 + b2
Y = sigmoid(Z2)

print(Y)
```

```python
# Assignment C — Vector Statistics (AI Version)
def vector_norm(x):
    return np.sqrt(np.sum(x**2))

def normalize_vector(x):
    return x / vector_norm(x)

def cosine_similarity(x, y):
    return np.dot(x, y) / (vector_norm(x) * vector_norm(y))

x = np.array([3, -4])
y = np.array([3, 2])
print(vector_norm(x))   # 5.0
print(normalize_vector(x))
print(cosine_similarity(x, y))

x = np.random.randn(512)
y = np.random.randn(512)

print("‖x‖ =", vector_norm(x))
print("‖normalize(x)‖ =", vector_norm(normalize_vector(x)))
print("cos(x, y) =", cosine_similarity(x, y))
```

```python
# Assignment D — Data Matrix Analytics
data = np.random.randn(4, 6)   # 4 samples, 6 features each

print(F"Mean: {np.mean(data, axis=0)}")
print(f"Variance: {np.var(data, axis=0)}")
print(f"L2 norm: {np.linalg.norm(data, axis=0)}")
print(f"Normalize each sample to unit norm: {data / np.linalg.norm(data, axis=1, keepdims=True)}")
print(f"Correlation matrix using np.corrcoef(data): {np.corrcoef(data, rowvar=False)}")
```

```python
import numpy as np

data = np.random.randn(4, 6)

# 1. Mean per feature
mean_per_feature = np.mean(data, axis=0)

# 2. Variance per feature
var_per_feature = np.var(data, axis=0)

# 3. L2 norm per sample
l2_per_sample = np.linalg.norm(data, axis=1)

# 4. Normalize each sample to unit norm
normalized_data = data / np.linalg.norm(data, axis=1, keepdims=True)

# 5. Correlation matrix
corr_matrix = np.corrcoef(data, rowvar=False)

print("Mean per feature:\n", mean_per_feature)
print("\nVariance per feature:\n", var_per_feature)
print("\nL2 norm per sample:\n", l2_per_sample)
print("\nNormalized data:\n", normalized_data)
print("\nCorrelation matrix:\n", corr_matrix)
```
