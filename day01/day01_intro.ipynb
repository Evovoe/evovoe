```python
# 3. Challenge (optional + recommended)
# Write a loop that prints numbers 1–20, but replaces:
"""
| Number             | Print Instead |
| ------------------ | ------------- |
| divisible by 3     | "Vision"      |
| divisible by 5     | "Game AI"     |
| divisible by 3 & 5 | "Chatbot"     |
"""
for i in range(1, 21):
    if (i % 3 == 0 and i % 5 == 0):
        print("Chatbot")
    elif (i % 3 == 0):
        print("Vision")
    elif (i % 5 == 0):
        print("Game AI")
    else:
        print(i)
```
